```python
df = pd.read_csv("SampleSuperstore.csv")
```

```python
print("Dataset shape:", df.shape)
```

```python
df_cleaned = df.drop_duplicates().copy()
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
print("Duplicates removed:", len(df) - len(df_cleaned))
```

```python
df_cleaned.dtypes
```

```python
data_type_summary = pd.DataFrame({
    "Column": df_cleaned.columns,
    "Data Type": df_cleaned.dtypes.astype(str).values
})

data_type_summary
```

```python
expected_numeric = ["Postal Code", "Sales", "Quantity", "Discount", "Profit"]

for column in expected_numeric:
    print(f"{column}: {df_cleaned[column].dtype}")
```

```python
expected_categorical = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Category",
    "Sub-Category"
]

for column in expected_categorical:
    print(f"{column}: {df_cleaned[column].dtype}")
```

```python
categorical_columns = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Category",
    "Sub-Category"
]

for column in categorical_columns:
    print("\n==============================")
    print(column)
    print("==============================")
    print(df_cleaned[column].unique())
```

```python


for column in categorical_columns:
    leading_trailing_spaces = (
        df_cleaned[column].astype(str) !=
        df_cleaned[column].astype(str).str.strip()
    ).sum()

    print(f"{column}: {leading_trailing_spaces} values with extra spaces")
```

```python
print("Quantity <= 0:", (df_cleaned["Quantity"] <= 0).sum())
print("Sales < 0:", (df_cleaned["Sales"] < 0).sum())
print("Discount < 0:", (df_cleaned["Discount"] < 0).sum())
print("Discount > 1:", (df_cleaned["Discount"] > 1).sum())

#profit Range
print("Minimum Profit:", df_cleaned["Profit"].min())
print("Maximum Profit:", df_cleaned["Profit"].max())

#Check Sales and Quantity ranges
print("Minimum Sales:", df_cleaned["Sales"].min())
print("Maximum Sales:", df_cleaned["Sales"].max())

print("Minimum Quantity:", df_cleaned["Quantity"].min())
print("Maximum Quantity:", df_cleaned["Quantity"].max())
```

```python
print("========== FINAL DATA QUALITY REPORT ==========")

print("Original rows:", df.shape[0])
print("Final rows:", df_cleaned.shape[0])
print("Columns:", df_cleaned.shape[1])

print("\nMissing values:")
print(df_cleaned.isnull().sum().sum())

print("\nDuplicate records:")
print(df_cleaned.duplicated().sum())

print("\nData types:")
print(df_cleaned.dtypes)
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Records"
    ],
    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        df_cleaned.shape[0],
        df_cleaned.shape[1],
        df_cleaned.isnull().sum().sum(),
        df_cleaned.duplicated().sum()
    ]
})

comparison
```

```python
df_cleaned.to_csv("cleaned_superstore.csv", index=False)

print("Cleaned dataset exported successfully!")
```
