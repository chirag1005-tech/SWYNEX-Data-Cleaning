```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("SampleSuperstore.csv")
print("Dataset loaded successfully!")
```

```python
df.head()
df.shape
```

```python
df.columns
```

```python
df.info
```

```python
df.describe()
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_summary
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
```

```python
duplicates = df[df.duplicated(keep=False)]

duplicates
```

```python
df_cleaned = df.drop_duplicates().copy()
```

```python
print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
print("Duplicates removed:", len(df) - len(df_cleaned))
```

```python
print("Remaining duplicates:", df_cleaned.duplicated().sum())
```

```python
df_cleaned.dtypes
```

```python
numeric_columns = [
    "Postal Code",
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for column in numeric_columns:
    print(column, ":", df_cleaned[column].dtype)
```

```python
categorical_columns = [
    "Ship Mode",
    "Segment",
    "Country",
    "City",
    "State",
    "Region",
    "Category",
    "Sub-Category"
]

for column in categorical_columns:
    print(column, ":", df_cleaned[column].dtype)
```

```python
for column in categorical_columns:
    print("\n---", column, "---")
    print(df_cleaned[column].unique())
```

```python
for column in categorical_columns:
    spaces = df_cleaned[column].astype(str).str.strip()
    if not df_cleaned[column].astype(str).equals(spaces):
        print("Spaces found in:", column)
```

```python
for column in categorical_columns:
    spaces = df_cleaned[column].astype(str).str.strip()
    if not df_cleaned[column].astype(str).equals(spaces):
        print("Spaces found in:", column)
```

```python
df_cleaned["Ship Mode"].value_counts()
df_cleaned["Segment"].value_counts()
df_cleaned["Region"].value_counts()
df_cleaned["Category"].value_counts()
df_cleaned["Sub-Category"].value_counts()
```

```python
print("Quantity <= 0:", (df_cleaned["Quantity"] <= 0).sum())
print("Sales < 0:", (df_cleaned["Sales"] < 0).sum())
print("Discount < 0:", (df_cleaned["Discount"] < 0).sum())
```

```python
print(df_cleaned["Postal Code"].dtype)
print(df_cleaned["Postal Code"].isnull().sum())
```

```python
print("===== FINAL DATA QUALITY CHECK =====")

print("Rows:", df_cleaned.shape[0])
print("Columns:", df_cleaned.shape[1])
print("Missing values:", df_cleaned.isnull().sum().sum())
print("Duplicate rows:", df_cleaned.duplicated().sum())
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Missing Values",
        "Duplicate Records"
    ],
    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        df_cleaned.shape[0],
        df_cleaned.shape[1],
        df_cleaned.isnull().sum().sum(),
        df_cleaned.duplicated().sum()
    ]
})

comparison
```

```python
df_cleaned.to_csv("cleaned_superstore.csv", index=False)

print("Cleaned dataset saved successfully!")
```
